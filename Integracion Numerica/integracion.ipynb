## Midpoint Rule

Lo siguiente corresponde a un algortimo especifico para poder resolver una integral ya definida.

```python
# Metodo de integracion Midpoint Rule:

lim = np.linspace(0,3,1000) #limite de integracion
int = 0.0

def midpoint_simple(f, a, b):
    # Funcion Midpoint que calcula integrales, donde f corresponde a la funcion a integrar y a,b el intervalo
    # de integracion
    return (b-a)*f((a+b)/2)

def g(x):
    #Corresponde a la funcion que integraremos
    return (x-1)*np.exp(x)



for i in range(len(lim)-1):
    int = int + midpoint_simple(g, lim[i], lim[i+1])
# Con el ciclo for iteramos de modo que nos acercamos al valor analitico
I = int 

print(I)    
```
